# Greek Letters and Jupyter Notebook Embedding

This notebook shows how to use Greek letters and LaTeX math in Markdown cells, and how to include notebooks in a GitHub `README`. It is intentionally simple and focused on documentation practice.

## 1. Displaying Greek Letters in Jupyter Notebooks

You can display Greek letters directly in Markdown using Unicode characters like α, β, γ, δ, and θ.

Example:
- Alpha: α
- Beta: β
- Gamma: γ
- Delta: δ
- Theta: θ

You can also use Greek letters in code output, for example by printing them from a Python string.

````python
# Print Greek letters from a Python string
print('Alpha: α')
print('Beta: β')
print('Gamma: γ')
print('Delta: δ')
print('Theta: θ')
````

## 2. Using LaTeX for Mathematical Expressions

You can write LaTeX inside Markdown cells using `$...$` for inline math or `$$...$$` for display math.

Examples:

Inline math: $\alpha + \beta = \gamma$  
Display math:

$$
\alpha^2 + \beta^2 = \gamma^2
$$

The notebook will render these equations automatically.

````python
# Example values for Greek-letter math
alpha = 1.5
beta = 2.0
gamma = alpha + beta
print(f"alpha = {alpha}, beta = {beta}, gamma = {gamma}")
````

## 3. Embedding Notebooks in GitHub README

To include a notebook in your GitHub repository README:

1. Push the notebook file (for example, `notebooks/greek_readme_demo.ipynb`) to GitHub.
2. GitHub will render the notebook automatically when browsing the file.
3. You can link to the notebook from `README.md`:

```markdown
[View the notebook](notebooks/greek_readme_demo.ipynb)
```

If you want a nicer rendered preview, use nbviewer:

```markdown
[View notebook on nbviewer](https://nbviewer.org/github/<your-username>/<your-repo>/blob/main/notebooks/greek_readme_demo.ipynb)
```

Replace `<your-username>` and `<your-repo>` with your GitHub details.
